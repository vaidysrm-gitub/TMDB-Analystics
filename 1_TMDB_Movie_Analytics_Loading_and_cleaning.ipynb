# movies

```python
df = pd.read_csv("/content/movies.csv")
df
```

```python
df.info()
```

```python
df.describe()
```

```python
df.describe(include="object")
```

```python
print(df.head())
print("Total rows:", len(df))
print("Column names:", df.columns.tolist())
print(df.dtypes)
print("Rows and columns:", df.shape)
```

```python
print(df.isnull().sum())
```

```python
df["release_date"] = pd.to_datetime(
    df["release_date"]

)

print("Missing dates:", df["release_date"].isnull().sum())
```

```python
df.info()
```

```python
df["release_date"]
```

```python
print(df.isnull())
```

```python
print(df.isnull().sum())
```

```python
df = pd.read_csv("movies.csv")

print("Duplicate rows:", df.duplicated().sum())
```

```python
duplicates = df[df.duplicated(keep=False)]

print(duplicates)
```

```python
df = df.drop_duplicates()

print("Duplicates remaining:", df.duplicated().sum())
```

```python
budget = df.loc[df["budget"] > 0, "budget"].mean()

revenue = df.loc[df["revenue"] > 0, "revenue"].mean()
print("Mean budget:", budget)
print("Mean revenue:", revenue)
```

```python
df.loc[df["budget"] == 0, "budget"] = budget
df.loc[df["revenue"] == 0, "revenue"] = revenue
```

```python
from google.colab import files
df.to_csv("movies_1.csv", index=False)

print("File saved successfully!")

files.download("movies_1.csv")
```
