# **3. List the profit (revenue − budget) and return on investment for every movie that has valid budget and revenue figures.**

```python
#4. Display movies with valid budget and revenue
valid_movies = movies[
    (movies["budget"] > 0) &
    (movies["revenue"] > 0)
].copy()

print("Movies with valid budget and revenue:",
      len(valid_movies))
```

```python
#Calculate profit
valid_movies["profit"] = (
    valid_movies["revenue"] - valid_movies["budget"]
)
print(valid_movies[["title", "budget", "revenue", "profit"]].head(10))
```

```python
#Calculate ROI
valid_movies["ROI"] = (
    valid_movies["profit"] / valid_movies["budget"]
) * 100
print(valid_movies[["title", "budget", "revenue", "profit", "ROI"]].head(10))
```

```python
from numpy._core.numeric import sort
#Display the results
valid_movies[
    ["movie_id", "title", "budget", "revenue", "profit", "ROI"]
]
valid_movies.sort_values(
    by="ROI",
    ascending=False
)

display(valid_movies.head(10))
```

```python
#Sort movies by ROI (highest to lowest)
sorted_movies = valid_movies.sort_values(
    by="ROI",
    ascending=False
)

# Step 2: Display top 10 movies
display(sorted_movies.head(10))
```

```python
#Additional analysis: Count profitable and loss-making movies
profitable_movies = valid_movies[
    valid_movies["profit"] > 0
]

loss_movies = valid_movies[
    valid_movies["profit"] < 0
]

break_even_movies = valid_movies[
    valid_movies["profit"] == 0
]

print("Profitable Movies:", len(profitable_movies))
print("Loss-making Movies:", len(loss_movies))
print("Break-even Movies:", len(break_even_movies))
```

# **6. List all movies whose runtime is a statistical outlier compared to the rest of the dataset.**

```python
print(movies["runtime"].describe())
```

```python
#Remove missing and zero runtime values
runtime_data = movies[
    movies["runtime"].notna() &
    (movies["runtime"] > 0)
].copy()
```

```python
Q1 = runtime_data["runtime"].quantile(0.25)

Q3 = runtime_data["runtime"].quantile(0.75)

IQR = Q3 - Q1

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
```

```python
lower_limit = Q1 - (1.5 * IQR)

upper_limit = Q3 + (1.5 * IQR)

print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)
```

```python
runtime_outliers = runtime_data[
    (runtime_data["runtime"] < lower_limit) |
    (runtime_data["runtime"] > upper_limit)
].copy()
```

```python
runtime_outliers = runtime_outliers.sort_values(
    by="runtime"
)

runtime_outliers[
    ["movie_id", "title", "release_date", "runtime"]
]
```

```python
print("Total Runtime Outlier Movies:", len(runtime_outliers))
```

# **8. List the top 10 most prolific actors by number of movies, along with their average movie rating across those movies.**

```python
actor_movies = cast[
    ["person_id", "actor_name", "movie_id"]
].copy()

print(actor_movies.head())
print("Total records:", len(actor_movies))
```

```python
#2. Find duplicate actor/movie combinations
duplicate_actors = actor_movies[
    actor_movies.duplicated(
        subset=["person_id", "movie_id"],
        keep=False
    )
]

print(duplicate_actors)
print("Duplicate rows:", len(duplicate_actors))
```

```python
#3. Remove duplicate combinations
actor_movies = actor_movies.drop_duplicates(
    subset=["person_id", "movie_id"],
    keep="first"
)

print("Records after removing duplicates:", len(actor_movies))
```

```python
#4. Add movie ratings
actor_movies = pd.merge(
    actor_movies,
    movies[["movie_id", "vote_average"]],
    on="movie_id",
    how="inner"
)

print(actor_movies.head())
```

```python
#5. Calculate the total number of movies for each actor
actor_count = actor_movies.groupby(
    ["person_id", "actor_name"]
)["movie_id"].nunique().reset_index()

actor_count = actor_count.rename(
    columns={"movie_id": "total_movies"}
)

actor_count.head()
```

```python
#6. Calculate the average movie rating
actor_rating = actor_movies.groupby(
    ["person_id", "actor_name"]
)["vote_average"].mean().reset_index()

actor_rating = actor_rating.rename(
    columns={"vote_average": "average_rating"}
)

actor_rating.head()
```

```python
#7. Combine movie count and average rating
actor_analysis = pd.merge(
    actor_count,
    actor_rating,
    on=["person_id", "actor_name"]
)

actor_analysis.head()
```

```python
#8. Sort actors by movie count
actor_analysis = actor_analysis.sort_values(
    by="total_movies",
    ascending=False
)

actor_analysis.head(10)
```

```python
# 9. Display the final top 10 actors
top_10_actors = actor_analysis.head(10).copy()

top_10_actors["average_rating"] = (
    top_10_actors["average_rating"].round(2)
)

top_10_actors
```

# **10. List movies whose popularity score and vote count don't follow the general pattern between the two — i.e., unusually high popularity with a low vote count, or vice versa.**

```python
popularity_data = movies[
    ["movie_id", "title", "popularity", "vote_count"]
].copy()

popularity_data.head()
```

```python
#remove missing values
popularity_data = popularity_data.dropna(
    subset=["popularity", "vote_count"]
)

print("Total movies:", len(popularity_data))
```

```python
#Calculate high and low values
#Consider 90th percentile = unusually high values.
#Consider 10th percentile = unusually low values.

high_popularity = popularity_data["popularity"].quantile(0.90)

low_popularity = popularity_data["popularity"].quantile(0.10)

high_votes = popularity_data["vote_count"].quantile(0.90)

low_votes = popularity_data["vote_count"].quantile(0.10)

print("High Popularity:", high_popularity)
print("Low Popularity:", low_popularity)
print("High Vote Count:", high_votes)
print("Low Vote Count:", low_votes)
```

```python
# Identify unusual movies
  # Condition 1: High popularity AND low vote count.
  # Condition 2: Low popularity AND high vote count.
unusual_movies = popularity_data[
    (
        (popularity_data["popularity"] >= high_popularity) &
        (popularity_data["vote_count"] <= low_votes)
    ) |
    (
        (popularity_data["popularity"] <= low_popularity) &
        (popularity_data["vote_count"] >= high_votes)
    )
].copy()
```

```python
#Display the unusual movies
unusual_movies = unusual_movies.sort_values(
    by="popularity",
    ascending=False
)

unusual_movies[
    ["movie_id", "title", "popularity", "vote_count"]
]
```

```python
# Print count of unusual movies
print("Total unusual movies:", len(unusual_movies))
```

```python
#Separate the two types of unusual movies
high_popularity_low_votes = popularity_data[
    (popularity_data["popularity"] >= high_popularity) &
    (popularity_data["vote_count"] <= low_votes)
]

low_popularity_high_votes = popularity_data[
    (popularity_data["popularity"] <= low_popularity) &
    (popularity_data["vote_count"] >= high_votes)
]

print("High popularity with low votes:",
      len(high_popularity_low_votes))

print("Low popularity with high votes:",
      len(low_popularity_high_votes))
```
